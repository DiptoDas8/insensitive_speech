## Formatting Hindu Catalogue

```python
hin_directory = 'data\Minority ARC data\Hindu\catalogue final.csv'
hin_data = pd.read_csv(hin_directory)
```

```python
# Select columns "Theme", "Comment" and "ExpertExplanation"
hin_data = hin_data[["Theme", "Comment", "ExpertsExplanation"]]
```

```python
# Select rows where "ExpertsExplanation" is not NaN
hin_withExplanations = hin_data.dropna(subset=["ExpertsExplanation"])
```

```python
# Comment with explanations
hin_withExplanations
```

```python
# Select rows from data where "Comment" is present but "ExpertsExplanation" is not
hin_withoutExplanations = hin_data[hin_data["ExpertsExplanation"].isnull() & hin_data["Comment"].notnull()]
```

```python
# Drop values in `Comment` that contain text "mp4"
hin_withoutExplanations = hin_withoutExplanations[~hin_withoutExplanations["Comment"].str.contains("mp4")]
```

```python
# Comment without explanations
hin_withoutExplanations
```

```python
# Tag the "Minority" in both dataframes as Hindu

hin_withExplanations.insert(0, "Minority", "Hindu")
hin_withoutExplanations.insert(0, "Minority", "Hindu")
```

## Formatting Chakma Catalogue

```python
chak_directory = 'data\Minority ARC data\Chakma\catalogue Final.csv'
chak_data = pd.read_csv(chak_directory)
```

```python
# Remove empty rows and replace newline characters
chak_data = chak_data.dropna(subset=["Comment", "ExpertsExplanation"]).replace(r'\n', ' ', regex=True).reset_index(drop=True)
```

```python
# Add a new column named "Minority" and set it to "Chakma"
chak_data.insert(0, "Minority", "Chakma")
```

```python
chak_data
```

```python
# Randomly choose 8 rows from chak_data and remove them from the dataframe and reset index
chak_withoutExplanations = chak_data.sample(8).reset_index(drop=True)
chak_withExplanations = chak_data.drop(chak_withoutExplanations.index).reset_index(drop=True)
```

```python
chak_withExplanations
```

```python
chak_withoutExplanations
```

```python
# Combine chak_withoutExplanations with hin_withoutExplanations
total_withoutExplanations = pd.concat([chak_withoutExplanations, hin_withoutExplanations]).reset_index(drop=True)
```

```python
# The goal of total_withoutExplanations is to have a test set of 8 rows of Chakma and 8 rows of Hindu of unseen data
total_withoutExplanations = total_withoutExplanations[["Minority", "Comment"]]
```

```python
total_withoutExplanations
```

```python
total_withExplanations = pd.concat([chak_withExplanations, hin_withExplanations]).reset_index(drop=True)
```

```python
total_withExplanations
```

```python
total_withExplanations[["Minority", "Comment", "ExpertsExplanation"]].to_csv("all_prompts.csv", index=False, encoding='utf-8-sig')
```
